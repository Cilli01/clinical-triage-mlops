```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
labels_path = "./data/raw/medical_tc_labels.csv"
test_path = "./data/raw/medical_tc_test.csv"
train_path = "./data/raw/medical_tc_train.csv"
```

```python
def plot_analysis(df):
    plt.figure(figsize=(5, 2))
    sns.countplot(data=df, x='urgency', order=['normal', 'attention', 'urgent'], palette='viridis')
    plt.title('Distribuição dos Níveis de Urgência para Triagem')
    plt.xlabel('Categoria de Triagem')
    plt.ylabel('Quantidade de Laudos')
    plt.show()
```

```python
from src.data.preprocess import preprocess_data

train_output_path = './data/processed/medical_tc_train_processed.csv'
df_train_formated = preprocess_data(train_path, train_output_path)

plot_analysis(df_train_formated)
```

```python
test_output_path = './data/processed/medical_tc_test_processed.csv'
df_test_formated = preprocess_data(test_path, test_output_path)

plot_analysis(df_test_formated)
```
